```python
#Endy Resta Febrian
#Melakukan penetapan class
class Endy_kalkulator_statistik:
    def __init__(self):
        self.Endy_data_base = {"Endy_data_set_1": [],
                               "Endy_data_set_2": []}

    def Endy_masukkan_data(self, Endy_jenis_data, Endy_data):
        self.Endy_data_base[Endy_jenis_data].append(Endy_data)

    def Endy_jumlah_data(self, Endy_jenis_data):
        Endy_jumlah_data = 0
        for Endy_i in self.Endy_data_base[Endy_jenis_data]:
            Endy_jumlah_data += Endy_i
        return Endy_jumlah_data

    def Endy_hapus_data(self, Endy_jenis_data, Endy_data):
        self.Endy_data_base[Endy_jenis_data].remove(Endy_data)

    def Endy_panjang_data(self, Endy_jenis_data):
        Endy_panjang_data = len(self.Endy_data_base[Endy_jenis_data])
        return Endy_panjang_data

    def Endy_rataan(self, Endy_jenis_data):
        Endy_rata_rata = self.Endy_jumlah_data(Endy_jenis_data) / self.Endy_panjang_data(Endy_jenis_data)
        return Endy_rata_rata

    def Endy_median(self, Endy_jenis_data):
        Endy_terurut = sorted(self.Endy_data_base[Endy_jenis_data])
        
        if len(Endy_terurut) % 2 != 0:
            Endy_median_ganjil = len(Endy_terurut) // 2
            Endy_median = Endy_terurut[Endy_median_ganjil]

        else:
            Endy_genap1 = len(Endy_terurut) // 2
            Endy_genap2 = len(Endy_terurut) // 2 - 1
            Endy_median = (Endy_terurut[Endy_genap1] + Endy_terurut[Endy_genap2]) / 2
        return Endy_median

    def Endy_range(self, Endy_jenis_data):
        Endy_range = max(self.Endy_data_base[Endy_jenis_data]) - min(self.Endy_data_base[Endy_jenis_data])
        return Endy_range
         
    def Endy_kuartil(self, Endy_jenis_data, Endy_q):
        if self.Endy_panjang_data(Endy_jenis_data) < 4:
            print("Panjang data minimal 4 agar dapat menghitung kuartil")
            return 0
        
        Endy_terurut = sorted(self.Endy_data_base[Endy_jenis_data])
        Endy_posisi = (Endy_q * (self.Endy_panjang_data(Endy_jenis_data) + 1)) / 4

        if Endy_posisi % 1 != 0:
            Endy_bawah = int(Endy_posisi)
            Endy_atas = Endy_bawah + 1

            Endy_data_bawah = Endy_terurut[Endy_bawah - 1]
            Endy_data_atas = Endy_terurut[Endy_atas - 1]

            Endy_desimal = Endy_posisi - Endy_bawah

            Endy_hasil = Endy_data_bawah + (Endy_desimal * (Endy_data_atas - Endy_data_bawah))

            return Endy_hasil 
             
        else:
            Endy_hasil = Endy_terurut[int(Endy_posisi) - 1]
            return Endy_hasil
             
    def Endy_kuartil_semua(self, Endy_jenis_data):
        Endy_semua_kuartil = []

        for Endy_q in range(1, 4):
            Endy_hasil = self.Endy_kuartil(Endy_jenis_data, Endy_q)
            Endy_semua_kuartil.append(Endy_hasil)

        return Endy_semua_kuartil

    def Endy_IQR(self, Endy_jenis_data):
        Endy_q3 = self.Endy_kuartil(Endy_jenis_data, 3)
        Endy_q1 = self.Endy_kuartil(Endy_jenis_data, 1)

        Endy_inter_kuartil = Endy_q3 - Endy_q1
        return Endy_inter_kuartil

    def Endy_varian(self, Endy_jenis_data):
        Endy_varian = 0
        Endy_rata_rata = self.Endy_rataan(Endy_jenis_data)
        
        for Endy_i in self.Endy_data_base[Endy_jenis_data]:
            Endy_varian += (Endy_i - Endy_rata_rata) ** 2
        
        Endy_varian /= self.Endy_panjang_data(Endy_jenis_data)
        return Endy_varian

    def Endy_standar_deviasi(self, Endy_jenis_data):
        return self.Endy_varian(Endy_jenis_data) ** 0.5

    def Endy_summary(self, Endy_jenis_data):
        Endy_summary_stat = {
            "Mean"           : self.Endy_rataan(Endy_jenis_data),
            "Median"         : self.Endy_median(Endy_jenis_data),
            "Range"          : self.Endy_range(Endy_jenis_data),
            "Quartil"        : self.Endy_kuartil_semua(Endy_jenis_data),
            "IQR"            : self.Endy_IQR(Endy_jenis_data),
            "Varian"         : self.Endy_varian(Endy_jenis_data),
            "Standar Deviasi": self.Endy_standar_deviasi(Endy_jenis_data)
        }

        for Endy_key, Endy_value in Endy_summary_stat.items():
            print(f"{Endy_key}: {Endy_value}")

        return Endy_summary_stat

    def Endy_summary_variabel(self, Endy_nilai_x, Endy_nilai_y):
        Endy_data_x = self.Endy_data_base[Endy_nilai_x]
        Endy_data_y = self.Endy_data_base[Endy_nilai_y]
        Endy_i = 0
        Endy_n = len(Endy_data_x)

        Endy_sigma_x = 0
        Endy_sigma_y = 0
        Endy_sigma_xy = 0
        Endy_sigma_x2 = 0
        Endy_sigma_y2 = 0

        for Endy_x in Endy_data_x:
            Endy_sigma_x += Endy_x

        for Endy_y in Endy_data_y:
            Endy_sigma_y += Endy_y

        for Endy_x in Endy_data_x:
            Endy_y = Endy_data_y[Endy_i]
            Endy_sigma_xy += Endy_x * Endy_y
            Endy_i += 1

        for Endy_x2 in Endy_data_x:
            Endy_sigma_x2 += Endy_x2 ** 2

        for Endy_y2 in Endy_data_y:
            Endy_sigma_y2 += Endy_y2 ** 2

        return Endy_sigma_x, Endy_sigma_y, Endy_sigma_xy, Endy_sigma_x2, Endy_sigma_y2, Endy_n

    def Endy_korelasi(self, Endy_nilai_x, Endy_nilai_y):
        Endy_sigma_x, Endy_sigma_y, Endy_sigma_xy, Endy_sigma_x2, Endy_sigma_y2, Endy_n = self.Endy_summary_variabel(Endy_nilai_x, Endy_nilai_y)
        
        Endy_bagian_atas = (Endy_n * Endy_sigma_xy) - (Endy_sigma_x * Endy_sigma_y)
        Endy_bagian_bawah_1 = (Endy_n * Endy_sigma_x2) - (Endy_sigma_x ** 2)
        Endy_bagian_bawah_2 = (Endy_n * Endy_sigma_y2) - (Endy_sigma_y ** 2)
        Endy_bagian_bawah = (Endy_bagian_bawah_1 * Endy_bagian_bawah_2) ** 0.5

        if Endy_bagian_bawah == 0:
            Endy_hasil_akhir = 0

        else:
            Endy_hasil_akhir = Endy_bagian_atas / Endy_bagian_bawah

        return Endy_hasil_akhir
        
    def Endy_model_regresi(self, Endy_nilai_x, Endy_nilai_y):
        Endy_sigma_x, Endy_sigma_y, Endy_sigma_xy, Endy_sigma_x2, Endy_sigma_y2, Endy_n = self.Endy_summary_variabel(Endy_nilai_x, Endy_nilai_y)

        Endy_atas_b = (Endy_n * Endy_sigma_xy) - (Endy_sigma_x * Endy_sigma_y)
        Endy_bawah_b = (Endy_n * Endy_sigma_x2) - (Endy_sigma_x ** 2)
        
        if Endy_bawah_b != 0:
            Endy_b = Endy_atas_b / Endy_bawah_b
            
        else:
            Endy_b = 0

        Endy_atas_a = Endy_sigma_y - (Endy_b * Endy_sigma_x)
        Endy_a = Endy_atas_a / Endy_n

        return Endy_a, Endy_b

    def Endy_determinasi(self, Endy_nilai_x, Endy_nilai_y):
        Endy_determinasi = self.Endy_korelasi(Endy_nilai_x, Endy_nilai_y) ** 2
        return Endy_determinasi

#Melakukan looping untuk mengakses class
Endy_kalstat = Endy_kalkulator_statistik()
while True:
    print("=" * 15, "Kalkulator Statistik", "=" * 15)
    print("[0.] Menampilkan data base")
    print("[1.] Menambahkan data")
    print("[2.] Menghapus data")
    print("[3.] Menampilkan ukuran pemusatan data")
    print("[4.] Menampilkan ukuran penyebaran data")
    print("[5.] Menampilkan summary statistics")
    print("[6.] Menampilkan korelasi pearson")
    print("[7.] Menampilkan model regresi linear sederhana")
    print("[8.] Menampilkan koefisien determinasi")
    print("[9.] Terimakasih, sistem akan berakhir")
    
    try:
        Endy_pilihan = int(input("Masukkan pilihan anda: "))
        print()
    
        if Endy_pilihan == 0:
            print("=" * 50)
            print("=" * 14, "Menampilkan Data Base", "=" * 15)
            print("=" * 50)
            print()
    
            print("=" * 8, "Pilih Data Set", "=" * 8)
            print("[1.] Data set 1")
            print("[2.] Data set 2")
            print("[3.] Keduanya")

            try:
                Endy_pilihan_tampil = int(input("Masukkan Pilihan Anda: "))
                print()
    
                if Endy_pilihan_tampil == 1:
                    Endy_isi = Endy_kalstat.Endy_data_base["Endy_data_set_1"]
                    if len(Endy_isi) == 0:
                        print("Data set 1 anda masih kosong, silahkan masukkan data anda!")
                        print()
                        continue
                    else:
                        print(f"Isi Data Set 1             : {Endy_isi}")
                        print(f"Banyak Data Pada Set 1 Anda: {len(Endy_isi)} ")
                        print()
        
                elif Endy_pilihan_tampil == 2:
                    Endy_isi = Endy_kalstat.Endy_data_base["Endy_data_set_2"]
                    if len(Endy_isi) == 0:
                        print("Data Set 2 anda masih kosong, silahkan masukkan data anda!")
                        print()
                        continue
                    else:
                        print(f"Isi Data Set 2 anda        : {Endy_isi}")
                        print(f"Banyak Data Pada Set 2 Anda: {len(Endy_isi)}")
                        print()

                elif Endy_pilihan_tampil == 3:
                    print("=" * 8, "Tampilan Kedua Data Set")
                    for Endy_key, Endy_value in Endy_kalstat.Endy_data_base.items():
                        if len(Endy_value) > 0:
                            Endy_tampilkan_data = Endy_value
                        else:
                            Endy_tampilkan_data = "KOSONG"
        
                        print(f"{Endy_key}: {Endy_tampilkan_data} dengan panjang {len(Endy_value)} data")
                        print()
        
                else:
                    print("Silahkan pilih pilihan 1/2/3 saja!!")
                    print()
                    continue

            except ValueError:
                print("Hanya masukkan angka saja!!")
                print()
                continue
    
        elif Endy_pilihan == 1:
            print("=" * 50)
            print("=" * 13, "Proses Memasukkan Data", "=" * 13)
            print("=" * 50)
            print()
        
            print("=" * 8, "Pilih Data Set", "=" * 8)
            print("[1.] Data set 1")
            print("[2.] Data set 2")

            try:
                Endy_pilihan_data = int(input("Masukkan pilihan data set anda: "))
                print()
        
                if Endy_pilihan_data == 1:
                    Endy_kunci = "Endy_data_set_1"
                    
                elif Endy_pilihan_data == 2:
                    Endy_kunci = "Endy_data_set_2"
                    
                else:
                    print("Error: Masukkan Pilihan 1/2 Saja")
                    print()
                    continue
        
                while True:
                    print("=" * 8, "Menu Menambahkan Data", "=" * 8)
                    print("[1.] Menambahkan data")
                    print("[2.] Input data selesai")

                    try:
                        Endy_sub_pilihan = int(input("Masukkan pilihan data anda: "))
        
                        if Endy_sub_pilihan == 1:
                            try:
                                Endy_input_data = float(input("Masukkan angka anda: "))
                                Endy_kalstat.Endy_masukkan_data(Endy_kunci, Endy_input_data)
                                print(f"Data {Endy_input_data} berhasil ditambahkan")
                                print(f"{Endy_kunci} anda sekarang {Endy_kalstat.Endy_data_base[Endy_kunci]}")
                                print()

                            except ValueError:
                                print("Hanya Masukkan Angka Saja!!")
            
                        elif Endy_sub_pilihan == 2:
                            print("Input data selesai")
                            print()
                            break
            
                        else:
                            print("Masukkan pilihan dengan benar!")
                            print()

                    except ValueError:
                        print("Masukkan Data Saja!!")
                        print()
                        continue
                        
            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue
    
        elif Endy_pilihan == 2:
            print("=" * 50)
            print("=" * 14, "Proses Menghapus Data", "=" * 15)
            print("=" * 50)
            print()
        
            print("=" * 8, "Pilih Data Set", "=" * 8)
            print("[1.] Data set 1")
            print("[2.] Data set 2")

            try:
                Endy_pilihan_data = int(input("Masukkan pilihan data set anda: "))
                print()
        
                if Endy_pilihan_data == 1:
                    Endy_kunci = "Endy_data_set_1"
                    
                elif Endy_pilihan_data == 2:
                    Endy_kunci = "Endy_data_set_2"
        
                else:
                    print("Error: Masukkan Pilihan 1/2 Saja")
                    continue
        
                while True:
                    print("=" * 8, "Menu Menghapus angka", "=" * 8)
                    print("[1.] Menghapus data dalam data set")
                    print("[2.] Menghapus data selesai")

                    try:
                        Endy_sub_pilihan = int(input("Masukkan pilihan anda: "))
                        print()
        
                        if Endy_sub_pilihan == 1:
                            print(f"Isi {Endy_kunci} anda sekarang {Endy_kalstat.Endy_data_base[Endy_kunci]}")
                            try:
                                Endy_data_hapus = float(input("Masukkan angka yang ingin di hapus: "))
            
                                if Endy_data_hapus in Endy_kalstat.Endy_data_base[Endy_kunci]:
                                    Endy_kalstat.Endy_hapus_data(Endy_kunci, Endy_data_hapus)
                                    print(f"Data {Endy_data_hapus} berhasil dihapus")
                
                                else:
                                    print("Data tidak ditemukan, masukkan dengan benar!!")
                
                                print(f"Isi {Endy_kunci} terbaru adalah: {Endy_kalstat.Endy_data_base[Endy_kunci]}")
                                print()

                            except ValueError:
                                print("Hanya Masukkan Angka Saja!!")
                                print()
                                continue
            
                        elif Endy_sub_pilihan == 2:
                            print("Menghapus Data Selesai")
                            print()
                            break
            
                        else:
                            print("Masukkan Pilihan Dengan Benar!")
                            print()

                    except ValueError:
                        print("Hanya Masukkan Data Saja!!")
                        print()
                        continue

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue

        elif Endy_pilihan == 3:
            print("=" * 50)
            print("=" * 11, "Menampilkan Ukuran Pemusatan Data", "=" * 11)
            print("=" * 50)
            print()
            
            print("=" * 8, "Pilihan ukuran pemusatan data", "=" * 8)
            print("[1.] Mean")
            print("[2.] Median")
            print("[3.] Semuan Ukuran Pemusatan Data")

            try:
                Endy_pilihan_pusat = int(input("Pemusatan data yang ingin dilihat: "))
                print()
        
                print("=" * 8, "Pilih Data Set", "=" * 8)
                print("[1.] Data set 1")
                print("[2.] Data set 2")

                try:
                    Endy_pilih_set = int(input("Masukkan pilihan data set anda: "))
                    print()
            
                    if Endy_pilih_set == 1:
                        Endy_kunci = "Endy_data_set_1"
            
                    elif Endy_pilih_set == 2:
                        Endy_kunci = "Endy_data_set_2"
        
                    else:
                        print("Error: Masukkan Pilihan 1/2 Saja")
                        print()
                        continue
        
                    if Endy_kalstat.Endy_panjang_data(Endy_kunci) != 0:
                        if Endy_pilihan_pusat == 1:
                            print(f"Mean dari {Endy_kunci} adalah {Endy_kalstat.Endy_rataan(Endy_kunci)}")
            
                        elif Endy_pilihan_pusat == 2:
                            print(f"Median dari {Endy_kunci} adalah {Endy_kalstat.Endy_median(Endy_kunci)}")
            
                        elif Endy_pilihan_pusat == 3:
                            print(f"Mean dari {Endy_kunci} adalah  : {Endy_kalstat.Endy_rataan(Endy_kunci)}")
                            print(f"Median dari {Endy_kunci} adalah: {Endy_kalstat.Endy_median(Endy_kunci)}")
                        
                        else:
                            print("Masukkan pilihan pemusatan dengan benar!!")
            
                    else:
                        print("Data set anda masih kosong, silahkan masukkan data anda!!")
            
                    print()

                except ValueError:
                    print("Hanya Masukkan Angka Saja!!")
                    print()
                    continue

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue

        elif Endy_pilihan == 4:
            print("=" * 50)
            print("=" * 8, "Menampilkan Ukuran Penyebaran Data")
            print("=" * 50)
            print()
    
            print("=" * 10, "Pilihan Ukuran Penyebaran Data", "=" * 10)
            print("[1.] Range")
            print("[2.] Kuartil")
            print("[3.] Interquartil")
            print("[4.] Varians")
            print("[5.] Standar Deviasi")
            print("[6.] Semua Ukuran Penyebaran Data")

            try:
                Endy_pilihan_sebaran = int(input("Masukkan Pilihan Sebaran Data Anda: "))
                print()
        
                print("=" * 8, "Pilih Data Set", "=" * 8)
                print("[1.] Data Set 1")
                print("[2.] Data Set 2")

                try:
                    Endy_pilih_set = int(input("Masukkanpilihan data set anda: "))
                    print()
        
                    if Endy_pilih_set == 1:
                        Endy_kunci = "Endy_data_set_1"
                    elif Endy_pilih_set == 2:
                        Endy_kunci = "Endy_data_set_2"
                    else:
                        print("Silahkan pilih pilihan 1/2 saja!!")
                        print()
                        continue
            
                    if Endy_kalstat.Endy_panjang_data(Endy_kunci) != 0:
                        if Endy_pilihan_sebaran == 1:
                            Endy_jangkauan = Endy_kalstat.Endy_range(Endy_kunci)
                            print(f"Jangkauan dari {Endy_kunci} adalah {Endy_jangkauan}")
                            print()
        
                        elif Endy_pilihan_sebaran == 2:
                            if Endy_kalstat.Endy_panjang_data(Endy_kunci) < 4:
                                print("Panjang data harus minimal 4!!")
                                print()
                                continue
                        
                            print("Pilihan Kuartil")
                            print("[1.] Kuartil ke-1")
                            print("[2.] Kuartil ke-2")
                            print("[3.] Kuartil ke-3")
                            print("[4.] Semua Pilihan Kuartil")

                            try:
                                Endy_pilihan_kuartil = int(input("Masukkan pilihan kuartil anda: "))
                
                                if Endy_pilihan_kuartil == 1:
                                    Endy_kuartil_1 = Endy_kalstat.Endy_kuartil(Endy_kunci, 1)
                                    print(f"Kuartil ke-1 anda dari {Endy_kunci}: {Endy_kuartil_1}")
                                    print()
                
                                elif Endy_pilihan_kuartil == 2:
                                    Endy_kuartil_2 = Endy_kalstat.Endy_kuartil(Endy_kunci, 2)
                                    print(f"Kuartil ke-2 anda dari {Endy_kunci}: {Endy_kuartil_2}")
                                    print()
                
                                elif Endy_pilihan_kuartil == 3:
                                    Endy_kuartil_3 = Endy_kalstat.Endy_kuartil(Endy_kunci, 3)
                                    print(f"Kuartil ke-3 anda dari {Endy_kunci}: {Endy_kuartil_3}")
                                    print()
                
                                elif Endy_pilihan_kuartil == 4:
                                    Endy_semua_quartil = Endy_kalstat.Endy_kuartil_semua(Endy_kunci)
                                    print(f"Kuartil 1-3 anda dari {Endy_kunci}: {Endy_semua_quartil}")
                                    print()
        
                                else:
                                    print("Silahkan Pilih Pilihan Dengan Benar!!")
                                    print()

                            except ValueError:
                                print("Hanya Masukkan Angka Saja!!")
                                print()
                                continue
            
                        elif Endy_pilihan_sebaran == 3:
                            if Endy_kalstat.Endy_panjang_data(Endy_kunci) < 4:
                                print("Panjang data harus minimal 4!!")
                                print()
                                continue
                                
                            Endy_interquartil = Endy_kalstat.Endy_IQR(Endy_kunci)
                            print(f"Jarak Interquartil dari {Endy_kunci}: {Endy_interquartil}")
                            print()
        
                        elif Endy_pilihan_sebaran == 4:
                            Endy_varians = Endy_kalstat.Endy_varian(Endy_kunci)
                            print(f"Varians dari {Endy_kunci}: {Endy_varians}")
                            print()
            
                        elif Endy_pilihan_sebaran == 5:
                            Endy_sd = Endy_kalstat.Endy_standar_deviasi(Endy_kunci)
                            print(f"Standar Deviasi dari {Endy_kunci}: {Endy_sd}")
                            print()
            
                        elif Endy_pilihan_sebaran == 6:
                            if Endy_kalstat.Endy_panjang_data(Endy_kunci) < 4:
                                print("Panjang data harus minimal 4!!")
                                print()
                                continue
                                
                            Endy_semuanya = {
                                "Range": Endy_kalstat.Endy_range(Endy_kunci),
                                "Kuartil 1": Endy_kalstat.Endy_kuartil(Endy_kunci, 1),
                                "Kuartil 2": Endy_kalstat.Endy_kuartil(Endy_kunci, 2),
                                "Kuartil 3": Endy_kalstat.Endy_kuartil(Endy_kunci, 3),
                                "Interkuartil": Endy_kalstat.Endy_IQR(Endy_kunci),
                                "Varians": Endy_kalstat.Endy_varian(Endy_kunci),
                                "Standar Deviasi": Endy_kalstat.Endy_standar_deviasi(Endy_kunci)
                            }
        
                            for Endy_key, Endy_value in Endy_semuanya.items():
                                print(f"{Endy_key}: {Endy_value}")
                            print()
            
                        else:
                            print("Masukkan Pilihan Menu Sebaran Anda dengan benar!!")
            
                    else:
                        print("Data Anda Masih Kosong!!")
                        
                except ValueError:
                    print("Hanya Masukkan Angka Saja!!")
                    print()
                    continue

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue
    
        elif Endy_pilihan == 5:
            print("=" * 50)
            print("=" * 10, "Menampilkan Summary Statistics", "=" * 11)
            print("=" * 50)
            print()
    
            print("=" * 10, "Pilih Data Set", "=" * 10)
            print("[1.] Data set 1")
            print("[2.] Data set 2")

            try:
                Endy_pilih_set = int(input("Masukkan pilihan data set anda: "))
                print()
        
                if Endy_pilih_set == 1:
                    Endy_kunci = "Endy_data_set_1"
                elif Endy_pilih_set == 2:
                    Endy_kunci = "Endy_data_set_2"
                else:
                    print("Silahkan pilih pilihan 1/2!!")
                    print()
                    continue
        
                Endy_panjang = Endy_kalstat.Endy_panjang_data(Endy_kunci)
        
                if Endy_panjang == 0:
                    print("Data set anda masih kosong!!")
                    print()
                    continue
                elif Endy_panjang < 4:
                    print("Untuk summary statistics memerlukan minimal 4 data!!")
                    print()
                    continue
                else:
                    print(f"Summary Statistics dari {Endy_kunci}: ")
                    Endy_kalstat.Endy_summary(Endy_kunci)
                    print()

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue
    
        elif Endy_pilihan == 6:
            print("=" * 50)
            print("=" * 11, "Menampilkan Korelasi Pearson", "=" * 11)
            print("=" * 50)
            print()
    
            print("=" * 10, "Menentukan Variabel X", "=" * 10)
            print("[1.] Data set 1")
            print("[2.] Data set 2")

            try:
                Endy_pilih_x = int(input("Pilih data sebagai X: "))
                print()
        
                if Endy_pilih_x == 1:
                    Endy_kunci_x = "Endy_data_set_1"
                elif Endy_pilih_x == 2:
                    Endy_kunci_x = "Endy_data_set_2"
                else:
                    print("Silahkan pilih pilihan 1/2 saja")
                    print()
                    continue

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue
    
            print("=" * 10, "Menetukan Variabel Y", "=" * 10)
            print("[1.] Data set 1")
            print("[2.] Data set 2")

            try:
                Endy_pilih_y = int(input("Pilih data sebagai Y: "))
                print()
        
                if Endy_pilih_y == 1:
                    Endy_kunci_y = "Endy_data_set_1"
                elif Endy_pilih_y == 2:
                    Endy_kunci_y = "Endy_data_set_2"
                else:
                    print("Silahkan pilih pilihan 1/2 saja!!")
                    print()
                    continue

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue
    
            if Endy_kunci_x == Endy_kunci_y:
                print("Variabel X dan Y tidak boleh berasal dari data set yang sama!!")
                print()
                continue
    
            Endy_panjang_x = Endy_kalstat.Endy_panjang_data(Endy_kunci_x)
            Endy_panjang_y = Endy_kalstat.Endy_panjang_data(Endy_kunci_y)

            if Endy_panjang_x == 0:
                print("Data X tidak boleh kosong!!")
                print()
                continue
            elif Endy_panjang_y == 0:
                print("Data Y tidak boleh kosong!!")
                print()
                continue
    
            if Endy_panjang_x != Endy_panjang_y:
                print("Panjang data harus sama")
                print(f"Panjang data X: {Endy_panjang_x}")
                print(f"Panjang data Y: {Endy_panjang_y}")
                print()
                continue
    
            Endy_hasil_akhir = Endy_kalstat.Endy_korelasi(Endy_kunci_x, Endy_kunci_y)
            print(f"Korelasi Pearson yang diperoleh: {Endy_hasil_akhir}")
    
        elif Endy_pilihan == 7:
            print("=" * 50)
            print("=" * 4, "Menampilkan Model Regresi Linear Sederhana", "=" * 4)
            print("=" * 50)
            print()

            print("=" * 10, "Menuntukan Variabel X", "=" * 10)
            print("[1.] Data set 1")
            print("[2.] Data set 2")

            try:
                Endy_pilih_x = int(input("Pilih data sebagai X: "))
                print()
        
                if Endy_pilih_x == 1:
                    Endy_kunci_x = "Endy_data_set_1"
                elif Endy_pilih_x == 2:
                    Endy_kunci_x = "Endy_data_set_2"
                else:
                    print("Silahkan pilih pilihan 1/2 saja!!")
                    print()
                    continue

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue
    
            print("=" * 10, "Menentukan Variabel Y", "=" * 10)
            print("[1.] Data set 1")
            print("[2.] Data set 2")

            try:
                Endy_pilih_y = int(input("Pilih data sebagai Y: "))
                print()
        
                if Endy_pilih_y == 1:
                    Endy_kunci_y = "Endy_data_set_1"
                elif Endy_pilih_y == 2:
                    Endy_kunci_y = "Endy_data_set_2"
                else:
                    print("Silahkan pilih pilihan 1/2 saja!!")
                    print()
                    continue
        
                if Endy_kunci_x == Endy_kunci_y:
                    print("Variabel X dan Y tidak boleh berasal dari data set yang sama")
                    print()
                    continue

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue
    
            Endy_panjang_x = Endy_kalstat.Endy_panjang_data(Endy_kunci_x)
            Endy_panjang_y = Endy_kalstat.Endy_panjang_data(Endy_kunci_y)
    
            if Endy_panjang_x == 0:
                print("Data x tidak boleh kosong!!")
                print()
                continue
            elif Endy_panjang_y == 0:
                print("Data Y tidak boleh kosong!!")
                print()
                continue
    
            if Endy_panjang_x != Endy_panjang_y:
                print("Panjang data harus sama!!")
                print(f"Panjang data X: {Endy_panjang_x}")
                print(f"Panjang data Y: {Endy_panjang_y}")
                print()
                continue
    
            Endy_a, Endy_b = Endy_kalstat.Endy_model_regresi(Endy_kunci_x, Endy_kunci_y)
            print(f"Model regresi yang diperoleh: Y = {Endy_a} + {Endy_b}X")
    
        elif Endy_pilihan == 8:
            print("=" * 50)
            print("=" * 7, "Menampilkan Koefisien Determinasi", "=" * 8)
            print("=" * 50)
            print()
    
            print("=" * 10, "Menentukan Variabel X", "=" * 10)
            print("[1.] Data set 1")
            print("[2.] Data set 2")

            try:
                Endy_pilih_x = int(input("Pilih data sebagai X: "))
                print()
        
                if Endy_pilih_x == 1:
                    Endy_kunci_x = "Endy_data_set_1"
                elif Endy_pilih_x == 2:
                    Endy_kunci_x = "Endy_data_set_2"
                else:
                    print("Silahkan pilih pilihan 1/2 saja!!")
                    print()
                    continue

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue

            print("=" * 10, "Menentukan variabel Y", "=" * 10)
            print("[1.] Data set 1")
            print("[2.] Data set 2")

            try:
                Endy_pilih_y = int(input("Pilih data sebagai Y: "))
                print()
        
                if Endy_pilih_y == 1:
                    Endy_kunci_y = "Endy_data_set_1"
                elif Endy_pilih_y == 2:
                    Endy_kunci_y = "Endy_data_set_2"
                else:
                    print("Silahkan pilih pilihan 1/2 saja!!")
                    print()
                    continue
        
                if Endy_kunci_x == Endy_kunci_y:
                    print("Variabel X dan Y tidak boleh berasasl dari data set yang sama")
                    print()
                    continue

            except ValueError:
                print("Hanya Masukkan Angka Saja!!")
                print()
                continue
    
            Endy_panjang_x = Endy_kalstat.Endy_panjang_data(Endy_kunci_x)
            Endy_panjang_y = Endy_kalstat.Endy_panjang_data(Endy_kunci_y)
    
            if Endy_panjang_x == 0:
                print("Data X tidak boleh kosong!!")
                print()
                continue
            elif Endy_panjang_y == 0:
                print("Data Y tidak boleh kosong!!")
                print()
                continue
    
            if Endy_panjang_x != Endy_panjang_y :
                print("Panjang data harus sama!!")
                print(f"Panjang data X: {Endy_panjang_x}")
                print(f"Panjang data Y: {Endy_panjang_y}")
                print()
                continue
    
            Endy_hasil_akhir = Endy_kalstat.Endy_determinasi(Endy_kunci_x, Endy_kunci_y)
            print(f"Koefisien Determinasi yang diperoleh: {Endy_hasil_akhir}")
    
        elif Endy_pilihan == 9:
            print("=" * 15, "TERIMAKASIH", "=" * 15)
            break

    except ValueError:
        print("Pilihan yang dimasukkan harus berupa angka!!")
        print()
        continue
```
