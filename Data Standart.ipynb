# Import Packages


```python
import pandas as Endy_pd
import numpy as Endy_np
from fast_ml.feature_selection import get_duplicate_features
```

# Proses

```python
#Mengubah suatu kolom menjadi numerik
Endy_df = Endy_pd.DataFrame({'Point': ['200', '100']})
Endy_df.info()

Endy_df['Point'] = Endy_df['Point'].astype(int)
Endy_df.info()
```

```python
#Mengubah string menjadi date time
Endy_df1 = Endy_pd.DataFrame({'Tanggal': ['20220824', '20261226']})
Endy_df1.info()

Endy_df1['Tanggal'] = Endy_pd.to_datetime(Endy_df1['Tanggal'], format = '%Y%m%d')
Endy_df1.info()
```

```python
#Mengangani teks yang tidak konsisten
Endy_df2 = Endy_pd.DataFrame({'Menu': ['ICE TEA', 'Ice Tea', 'ice tea']})

#1. Mengubah menjadi kecil semua
Endy_df2['Menu'] = Endy_df2['Menu'].str.lower()
Endy_df2
```

```python
#2 Mengubah menjadi besar semua
Endy_df2['Menu'] = Endy_df2['Menu'].str.upper()
Endy_df2
```

```python
#3 Menjadi huruf besar diawal kata
Endy_df2['Menu'] = Endy_df2['Menu'].str.title()
Endy_df2
```

```python
#Menangani Spas yang tidak perlu
Endy_df3 = Endy_pd.DataFrame({'Menu': ['  Ice Tea  ', '  Ice Tea', 'Ice Tea  ']})
print(Endy_df3)

Endy_df3['Menu'] = Endy_df3['Menu'].str.strip()
print('\n')
print(Endy_df3)
```

```python
#Menghapus string yang tidak perlu
Endy_df4 = Endy_pd.DataFrame({'Produk':['ICE TEA (SOLD PER 1 GLASS)','RAINBOW CAKE (SOLD PER 1 SLICE)']})
print(Endy_df4)

Endy_df4['Produk'] = Endy_df4['Produk'].str.replace(r' \(SOLD.*', '', regex = True)
print(Endy_df4)
```

```python
#Mengubah sebuah string
Endy_df5 = Endy_pd.DataFrame({'Alamat': ['jl. Baru', 'jl. Indah']})
print(Endy_df5)

Endy_df5['Alamat'] = Endy_df5['Alamat'].str.replace(r'jl.* ', 'Jalan ', regex = True)
print(Endy_df5)
```
