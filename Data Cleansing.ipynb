# Import Data

```python
Endy_data = Endy_pd.DataFrame({'Visitor_name': ['Ivan', 'Denil', Endy_np.nan],
                             'Event_name': ['Rock Concert 2022', 'Jazz Festifal 2021', 'Pop Party 2024'],
                             'Ticket_sold_year': [2022, Endy_np.nan, 2024],
                             'Event_year': [2022, 2021, 2024],
                             'Ticket_type': ['Gold', 'Silver', 'Bronze'],
                             'Ticket_price': [1000000.0, 750000.0, 500000.0]})

Endy_data['Ticket_sold_year'] = Endy_data['Ticket_sold_year'].astype('Int64')
Endy_data['Event_year'] = Endy_data['Event_year'].astype('Int64')
Endy_data['Ticket_price'] = Endy_data['Ticket_price'].astype(float)
Endy_data
```

```python
Endy_data.info()
```

```python
Endy_data2 = Endy_pd.DataFrame({
    'Visitor_name': ['Andy', 'Ivan', 'Andhitya', 'Messi', 'Christ'],
    'Age': [25, Endy_np.nan, 32, 25, 19],
    'Event_name': ['Harmony Fest 2020', 'Skyline Music Show 2023', 'Vibrance Concert 2024', 'Zenith Music Festival 2025', 'Unity Sound Fest 2026'],
    'Ticket_sold_year': [2020, 2023, 2024, 2025, 2026],
    'Event_year': [2020, 2023, 2024, 2025, 2026],
    'Ticket_type': ['Gold', 'Bronze', 'Platinum', 'Gold', 'Silver'],
    'Ticket_price': [1000000.0, 500000.0, 2500000.0, 750000.0, 1000000.0],
    'Reward_point': [600, 300, 750, Endy_np.nan, 450]
})

Endy_data2['Age'] = Endy_data2['Age'].astype('Int64')
Endy_data2['Ticket_sold_year'] = Endy_data2['Ticket_sold_year'].astype('Int64')
Endy_data2['Event_year'] = Endy_data2['Event_year'].astype('Int64')
Endy_data2['Reward_point'] = Endy_data2['Reward_point'].astype('Int64')
Endy_data2['Ticket_price'] = Endy_data2['Ticket_price'].astype(float)
Endy_data2
```

```python
Endy_data2.info()
```

# Olah Data

### Hapus data

```python
#Jika kita ingin meng-drop yang ada NaN
Endy_data.dropna()
```

```python
#Menghapus kolom
Endy_data.drop(columns = ['Visitor_name', 'Ticket_sold_year'])
```

### Pengisian Data

```python
#Mengisi NA pada reward point berdasarkan informasi yang ada pada tabel
Endy_data2['Reward_point'] = Endy_data2['Reward_point'].fillna(600)
Endy_data2
```

```python
#Mengisi NA dengan Mean
Endy_data2_mean = Endy_data2.copy()
Endy_data2_mean['Age'] = Endy_data2_mean['Age'].fillna(Endy_data2_mean['Age'].mean().round())
Endy_data2_mean
```

```python
#Mengisi nilai NA dengan Median
Endy_data2_med = Endy_data2.copy()
Endy_data2_med['Age'] = Endy_data2_med['Age'].fillna(Endy_data2_med['Age'].median())
Endy_data2_med
```

```python
#Mengisi nilai NA dengan Modus
Endy_data2_mod = Endy_data2.copy()
Endy_data2_mod['Age'] = Endy_data2_mod['Age'].fillna(Endy_data2_mod['Age'].mode()[0])
Endy_data2_mod
```

```python
#Mengisi NA dengan nilai sebelumnya
Endy_data2_ffill = Endy_data2.copy()
Endy_data2_ffill['Age'] = Endy_data2_ffill['Age'].fillna(method = 'ffill')
Endy_data2_ffill
```

```python
#Mengisi Nilai NA dengan Nilai Setelahnya
Endy_data2_bfill = Endy_data2.copy()
Endy_data2_bfill['Age'] = Endy_data2_bfill['Age'].fillna(method = 'bfill')
Endy_data2_bfill
```
