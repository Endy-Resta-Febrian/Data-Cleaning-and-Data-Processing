```python
# Endy Resta Febrian
#Melakukan penetapan class
class Endy_kalkulator_statistik:
    def __init__(self):
        self.Endy_data_base = {}

    def Endy_buat_dataset(self, Endy_nama_dataset):
        if Endy_nama_dataset in self.Endy_data_base:
            print("Nama data set telah digunakan!!")
            return

        self.Endy_data_base[Endy_nama_dataset] = []
        print(f"Dataset {Endy_nama_dataset} berhasil ditambahkan")

    def Endy_hapus_dataset(self, Endy_nama_dataset):
        if Endy_nama_dataset not in self.Endy_data_base:
            print("Nama dataset tidak ada!!")
            return

        del self.Endy_data_base[Endy_nama_dataset]

    def Endy_tampilkan_dataset(self):
        if len(self.Endy_data_base) == 0:
            print("Data base masih kosong!! Masukkan dataset terlebih dahulu")
            return
        else:
            for Endy_nama_dataset in self.Endy_data_base:
                print(f"- {Endy_nama_dataset}")

    def Endy_masukkan_data(self, Endy_jenis_data, Endy_data):
        self.Endy_data_base[Endy_jenis_data].append(Endy_data)

    def Endy_jumlah_data(self, Endy_jenis_data):
        Endy_jumlah_data = 0
        for Endy_i in self.Endy_data_base[Endy_jenis_data]:
            Endy_jumlah_data += Endy_i
        return Endy_jumlah_data

    def Endy_hapus_data(self, Endy_jenis_data, Endy_data):
        self.Endy_data_base[Endy_jenis_data].remove(Endy_data)

    def Endy_panjang_data(self, Endy_jenis_data):
        Endy_panjang_data = len(self.Endy_data_base[Endy_jenis_data])
        return Endy_panjang_data

    def Endy_rataan(self, Endy_jenis_data):
        Endy_rata_rata = self.Endy_jumlah_data(Endy_jenis_data) / self.Endy_panjang_data(Endy_jenis_data)
        return Endy_rata_rata

    def Endy_median(self, Endy_jenis_data):
        Endy_terurut = sorted(self.Endy_data_base[Endy_jenis_data])
        
        if len(Endy_terurut) % 2 != 0:
            Endy_median_ganjil = len(Endy_terurut) // 2
            Endy_median = Endy_terurut[Endy_median_ganjil]

        else:
            Endy_genap1 = len(Endy_terurut) // 2
            Endy_genap2 = len(Endy_terurut) // 2 - 1
            Endy_median = (Endy_terurut[Endy_genap1] + Endy_terurut[Endy_genap2]) / 2
        return Endy_median

    def Endy_range(self, Endy_jenis_data):
        Endy_range = max(self.Endy_data_base[Endy_jenis_data]) - min(self.Endy_data_base[Endy_jenis_data])
        return Endy_range
         
    def Endy_kuartil(self, Endy_jenis_data, Endy_q):
        if self.Endy_panjang_data(Endy_jenis_data) < 4:
            print("Panjang data minimal 4 agar dapat menghitung kuartil")
            return "Isi Dataset Terlalu Pendek"
        
        Endy_terurut = sorted(self.Endy_data_base[Endy_jenis_data])
        Endy_posisi = (Endy_q * (self.Endy_panjang_data(Endy_jenis_data) + 1)) / 4

        if Endy_posisi % 1 != 0:
            Endy_bawah = int(Endy_posisi)
            Endy_atas = Endy_bawah + 1

            Endy_data_bawah = Endy_terurut[Endy_bawah - 1]
            Endy_data_atas = Endy_terurut[Endy_atas - 1]

            Endy_desimal = Endy_posisi - Endy_bawah

            Endy_hasil = Endy_data_bawah + (Endy_desimal * (Endy_data_atas - Endy_data_bawah))

            return Endy_hasil 
             
        else:
            Endy_hasil = Endy_terurut[int(Endy_posisi) - 1]
            return Endy_hasil
             
    def Endy_kuartil_semua(self, Endy_jenis_data):
        if self.Endy_panjang_data(Endy_jenis_data) < 4:
            return "Isi Dasteset Terlalu Pendek (minimal 4 data)"
            
        Endy_semua_kuartil = []

        for Endy_q in range(1, 4):
            Endy_hasil = self.Endy_kuartil(Endy_jenis_data, Endy_q)
            Endy_semua_kuartil.append(Endy_hasil)

        return Endy_semua_kuartil

    def Endy_IQR(self, Endy_jenis_data):
        if self.Endy_panjang_data(Endy_jenis_data) < 4:
            return "Isi Dataset Terlalu Pendek (minimal 4 data)"
            
        Endy_q3 = self.Endy_kuartil(Endy_jenis_data, 3)
        Endy_q1 = self.Endy_kuartil(Endy_jenis_data, 1)

        Endy_inter_kuartil = Endy_q3 - Endy_q1
        return Endy_inter_kuartil

    def Endy_varian(self, Endy_jenis_data):
        Endy_varian = 0
        Endy_rata_rata = self.Endy_rataan(Endy_jenis_data)
        
        for Endy_i in self.Endy_data_base[Endy_jenis_data]:
            Endy_varian += (Endy_i - Endy_rata_rata) ** 2
        
        Endy_varian /= self.Endy_panjang_data(Endy_jenis_data)
        return Endy_varian

    def Endy_standar_deviasi(self, Endy_jenis_data):
        return self.Endy_varian(Endy_jenis_data) ** 0.5

    def Endy_summary(self, Endy_jenis_data):
        Endy_summary_stat = {
            "Mean"           : self.Endy_rataan(Endy_jenis_data),
            "Median"         : self.Endy_median(Endy_jenis_data),
            "Range"          : self.Endy_range(Endy_jenis_data),
            "Quartil"        : self.Endy_kuartil_semua(Endy_jenis_data),
            "IQR"            : self.Endy_IQR(Endy_jenis_data),
            "Varian"         : self.Endy_varian(Endy_jenis_data),
            "Standar Deviasi": self.Endy_standar_deviasi(Endy_jenis_data)
        }

        for Endy_key, Endy_value in Endy_summary_stat.items():
            print(f"{Endy_key}: {Endy_value}")

        return Endy_summary_stat

    def Endy_summary_variabel(self, Endy_nilai_x, Endy_nilai_y):
        Endy_data_x = self.Endy_data_base[Endy_nilai_x]
        Endy_data_y = self.Endy_data_base[Endy_nilai_y]
        Endy_i = 0
        Endy_n = len(Endy_data_x)

        Endy_sigma_x = 0
        Endy_sigma_y = 0
        Endy_sigma_xy = 0
        Endy_sigma_x2 = 0
        Endy_sigma_y2 = 0

        for Endy_x in Endy_data_x:
            Endy_sigma_x += Endy_x

        for Endy_y in Endy_data_y:
            Endy_sigma_y += Endy_y

        for Endy_x in Endy_data_x:
            Endy_y = Endy_data_y[Endy_i]
            Endy_sigma_xy += Endy_x * Endy_y
            Endy_i += 1

        for Endy_x2 in Endy_data_x:
            Endy_sigma_x2 += Endy_x2 ** 2

        for Endy_y2 in Endy_data_y:
            Endy_sigma_y2 += Endy_y2 ** 2

        return Endy_sigma_x, Endy_sigma_y, Endy_sigma_xy, Endy_sigma_x2, Endy_sigma_y2, Endy_n

    def Endy_korelasi(self, Endy_nilai_x, Endy_nilai_y):
        Endy_sigma_x, Endy_sigma_y, Endy_sigma_xy, Endy_sigma_x2, Endy_sigma_y2, Endy_n = self.Endy_summary_variabel(Endy_nilai_x, Endy_nilai_y)
        
        Endy_bagian_atas = (Endy_n * Endy_sigma_xy) - (Endy_sigma_x * Endy_sigma_y)
        Endy_bagian_bawah_1 = (Endy_n * Endy_sigma_x2) - (Endy_sigma_x ** 2)
        Endy_bagian_bawah_2 = (Endy_n * Endy_sigma_y2) - (Endy_sigma_y ** 2)
        Endy_bagian_bawah = (Endy_bagian_bawah_1 * Endy_bagian_bawah_2) ** 0.5

        if Endy_bagian_bawah == 0:
            Endy_hasil_akhir = 0

        else:
            Endy_hasil_akhir = Endy_bagian_atas / Endy_bagian_bawah

        return Endy_hasil_akhir
        
    def Endy_model_regresi(self, Endy_nilai_x, Endy_nilai_y):
        Endy_sigma_x, Endy_sigma_y, Endy_sigma_xy, Endy_sigma_x2, Endy_sigma_y2, Endy_n = self.Endy_summary_variabel(Endy_nilai_x, Endy_nilai_y)

        Endy_atas_b = (Endy_n * Endy_sigma_xy) - (Endy_sigma_x * Endy_sigma_y)
        Endy_bawah_b = (Endy_n * Endy_sigma_x2) - (Endy_sigma_x ** 2)
        
        if Endy_bawah_b != 0:
            Endy_b = Endy_atas_b / Endy_bawah_b
            
        else:
            Endy_b = 0

        Endy_atas_a = Endy_sigma_y - (Endy_b * Endy_sigma_x)
        Endy_a = Endy_atas_a / Endy_n

        return Endy_a, Endy_b

    def Endy_determinasi(self, Endy_nilai_x, Endy_nilai_y):
        Endy_determinasi = self.Endy_korelasi(Endy_nilai_x, Endy_nilai_y) ** 2
        return Endy_determinasi

#Melakukan looping untuk mengakses class
Endy_kalstat = Endy_kalkulator_statistik()
while True:
    print("=" * 15, "Kalkulator Statistik", "=" * 15)
    print("[0.] Menampilkan data base")
    print("[1.] Menambahkan dataset")
    print("[2.] Menambahkan data")
    print("[3.] Menghapus dataset")
    print("[4.] Menghapus data")
    print("[5.] Menampilkan ukuran pemusatan data")
    print("[6.] Menampilkan ukuran penyebaran data")
    print("[7.] Menampilkan summary statistics")
    print("[8.] Menampilkan korelasi pearson")
    print("[9.] Menampilkan model regresi linear sederhana")
    print("[10.] Menampilkan koefisien determinasi")
    print("[11.] Terimakasih, sistem akan berakhir")
    
    try:
        Endy_pilihan = int(input("Masukkan pilihan anda: "))
        print()
    
        if Endy_pilihan == 0:
            print("=" * 50)
            print("=" * 13, "Menampilkan Data Base", "=" * 14)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Data base anda masih kosong!!")
                print()
                continue

            for Endy_key, Endy_value in Endy_kalstat.Endy_data_base.items():
                if len(Endy_value) == 0:
                    Endy_tampil = "Kosong"
                else:
                    Endy_tampil = Endy_value

                print(f"Nama dataset   : {Endy_key}")
                print(f"Isi Dataset    : {Endy_tampil}")
                print(f"Panjang Dataset: {len(Endy_value)}")
                print("-" * 50)

            print()

        elif Endy_pilihan == 1:
            print("=" * 50)
            print("=" * 14, "Menambahkan Dataset", "=" * 14)
            print("=" * 50)

            Endy_nama_dataset = input("Masukkan Nama Dataset Anda: ")
            if Endy_nama_dataset.strip() == "":
                print("Nama Dataset Tidak Boleh Kosong!!")
                print()
                continue
            else:
                Endy_kalstat.Endy_buat_dataset(Endy_nama_dataset)
                print()
    
        elif Endy_pilihan == 2:
            print("=" * 50)
            print("=" * 13, "Proses Memasukkan Data", "=" * 13)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Dataset Anda Masih Kosong!!")
                print()
                continue

            else:
                print("=" * 10, "Daftar Dataset", "=" * 10)

                for Endy_nama in Endy_kalstat.Endy_data_base:
                    print(f"- {Endy_nama}")

                print()
                Endy_pilih_dataset = input("Pilih Dataset yang ingin anda isi: ")
                print()

                if Endy_pilih_dataset not in Endy_kalstat.Endy_data_base:
                    print("Dataset tidak ditemukan!!")
                    print()
                    continue

                else:
                    while True:
                        print("Masukkan Data anda: ")
                        Endy_data = input("Ketika (selesai) jika sudah selesai").lower()

                        if Endy_data == "selesai":
                            print(f"Isi dari dataset {Endy_pilih_dataset}: {Endy_kalstat.Endy_data_base[Endy_pilih_dataset]}")
                            print()
                            break
                        else:
                            try:
                                Endy_data = float(Endy_data)
                                Endy_kalstat.Endy_masukkan_data(Endy_pilih_dataset, Endy_data)
                                
                                print(f"{Endy_data} berhasil ditambahkan")
                                print(f"Isi dataset {Endy_pilih_dataset}: {Endy_kalstat.Endy_data_base[Endy_pilih_dataset]}")
                                print("-" * 30)
                                print()
                                
                            except ValueError:
                                print("Mohon Ikuti Intruksi!!")
                                print()
                                continue
    
        elif Endy_pilihan == 3:
            print("=" * 50)
            print("=" * 12, "Proses Menghapus Dataset", "=" * 12)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Dataset Anda Masih Kosong!!")
                print()
                continue
            else:
                while True:
                    print("=" * 10, "Daftar Dataset", "=" * 10)
                    for Endy_nama in Endy_kalstat.Endy_data_base:
                        print(f"- {Endy_nama}")
    
                    print()
                    Endy_pilih_dataset = input("Masukkan nama dataset: ")
                    print()
    
                    if Endy_pilih_dataset not in Endy_kalstat.Endy_data_base:
                        print("Dataset tidak ditemukan!!")
                        print()
                        continue
                    else:
                        Endy_kalstat.Endy_hapus_dataset(Endy_pilih_dataset)
                        print(f"{Endy_pilih_dataset} berhasil dihapus")
                        print("Isi Database anda sekarang: ")
                        for Endy_nama_baru in Endy_kalstat.Endy_data_base:
                            print(f"- {Endy_nama_baru}")
                            print()
                        break

        elif Endy_pilihan == 4:
            print("=" * 50)
            print("=" * 14, "Menghapus Data", "=" * 15)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Data base anda masih kosong!!")
                print()
                continue
            else:
                print("Dataset anda sekarang:")
                for Endy_nama in Endy_kalstat.Endy_data_base:
                    print(f"- {Endy_nama}")
    
                print()
                Endy_pilih_dataset = input("Masukkan nama dataset anda")
                print()
    
                if Endy_pilih_dataset not in Endy_kalstat.Endy_data_base:
                    print("Dataset tidak ditemukan")
                    print()
                    continue
                else:
                    while True:
                        print(f"Isi Dataset {Endy_pilih_dataset}: {Endy_kalstat.Endy_data_base[Endy_pilih_dataset]}")
                        print("Pilih data yang ingin dihapus: ")
                        Endy_data = input("Ketik (selesai) jika sudah selesai").lower()
    
                        if Endy_data == "selesai":
                            print(f"Data set {Endy_pilih_dataset}: {Endy_kalstat.Endy_data_base[Endy_pilih_dataset]}")
                            print()
                            break
                        else:
                            try:
                                Endy_data = float(Endy_data)

                                if Endy_data not in Endy_kalstat.Endy_data_base[Endy_pilih_dataset]:
                                    print("Data tidak ditemukan")
                                    print("-" * 30)
                                    print()
                                    continue
                                else:
                                    Endy_kalstat.Endy_hapus_data(Endy_pilih_dataset, Endy_data)
                                    print(f"Data {Endy_data} berhasil dihapus dari {Endy_pilih_dataset}")
                                    print(f"Isi dari dataset {Endy_pilih_dataset}: {Endy_kalstat.Endy_data_base[Endy_pilih_dataset]}")
                                    print("-" * 30)
                                    print()
                                    
                            except ValueError:
                                print("Mohon Ikuti Instruksi!!")
                                print()
                                continue
                                
            
        elif Endy_pilihan == 5:
            print("=" * 50)
            print("=" * 8, "Menampilkan Ukuran Pemusatan Data", "=" * 9)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Data base anda masih kosong!!")
                print()
                continue
            else:
                print("=" * 10, "Daftar Dataset", "=" * 10)
                for Endy_nama in Endy_kalstat.Endy_data_base:
                    print(f"- {Endy_nama}")

                print()
                Endy_pilih_dataset = input("Masukkan nama dataset anda: ")
                print()

                if Endy_pilih_dataset not in Endy_kalstat.Endy_data_base:
                    print("Dataset tidak ditemukan!!")
                    print()
                    continue
                else:
                    print("[1.] Mean")
                    print("[2.] Median")
                    print("[3.] Kedunya")

                    while True:
                        Endy_sub_pilihan = input("Masukkan pilihan anda (selesai, jika sudah selesai): ").lower()
                        if Endy_sub_pilihan == "selesai":
                            print("Ukuran Pemusatan Data Selesai")
                            print("=" * 30)
                            print()
                            break
                        elif Endy_sub_pilihan == "1":
                            if Endy_kalstat.Endy_panjang_data(Endy_pilih_dataset) == 0:
                                print("Isi dataset anda masih kosong, mohon diisi terlebih dahulu!")
                                print()
                                break
                            else:
                                Endy_pilih_mean = Endy_kalstat.Endy_rataan(Endy_pilih_dataset)
                                print(f"Menampilkan Mean dari {Endy_pilih_dataset}: {Endy_pilih_mean}")
                                print("-" * 30)
                                print()
                                continue
                        elif Endy_sub_pilihan =="2":
                            if Endy_kalstat.Endy_panjang_data(Endy_pilih_dataset) == 0:
                                print("Isi dataset anda masih kosong, mohon diisi terlebih dahulu!")
                                print()
                                break
                            else:
                                Endy_pilih_median = Endy_kalstat.Endy_median(Endy_pilih_dataset)
                                print(f"Menampilkan Median dari {Endy_pilih_dataset}: {Endy_pilih_median}")
                                print("-" * 30)
                                print()
                                continue
                        elif Endy_sub_pilihan == "3":
                            if Endy_kalstat.Endy_panjang_data(Endy_pilih_dataset) == 0:
                                print("Isi dataset anda masih kosong, mohon diisi terlebih dahulu!")
                                print()
                                break
                            else:
                                Endy_pilih_mean = Endy_kalstat.Endy_rataan(Endy_pilih_dataset)
                                Endy_pilih_median = Endy_kalstat.Endy_median(Endy_pilih_dataset)

                                print(f"Menampilkan Mean   dari {Endy_pilih_dataset}: {Endy_pilih_mean}")
                                print(f"Menampilkan Median dari {Endy_pilih_dataset}: {Endy_pilih_median}")
                                print("-" * 30)
                                print()
                                continue
                        else:
                            print("Silahkan pilih pilihan yang tersedia atau ikuti instruksi!")
                            print()
                            continue

        elif Endy_pilihan == 6:
            print("=" * 50)
            print("=" * 7, "Menampilkan Ukuran Penyebaran Data", "=" * 7)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Data base anda masih kosong!!")
                print()
                continue
            else:
                while True:
                    print("=" * 10, "Daftar Dataset Anda", "=" * 10)
                    for Endy_nama in Endy_kalstat.Endy_data_base:
                        print(f"- {Endy_nama}")
                        
                    print()
                    Endy_pilih_dataset = input("Masukkan nama dataset / exit: ")
                    if Endy_pilih_dataset == "exit":
                        break
                    elif Endy_pilih_dataset not in Endy_kalstat.Endy_data_base:
                        print("Dataset tidak ditemukan!!")
                        print()
                        continue
                    else:
                        if Endy_kalstat.Endy_panjang_data(Endy_pilih_dataset) == 0:
                            print("Isi dataset anda masih kosong!")
                            print()
                            continue
                        else:
                            while True:
                                print("=" * 10, "Menu Penyebaran Data", "=" * 10)
                                print("[1.] Range")
                                print("[2.] Quartil")
                                print("[3.] IQR")
                                print("[4.] Varians")
                                print("[5.] Standar Deviasi")
                                print("[6.] All")
                                print()
                                
                                Endy_penyebaran = input("Masukkan pilihan anda / selesai: ")
                                if Endy_penyebaran == "selesai":
                                    print("Menu Selesai")
                                    print()
                                    break
                                elif Endy_penyebaran == "1":
                                    Endy_jangkauan = Endy_kalstat.Endy_range(Endy_pilih_dataset)
                                    print(f"Menampilkan Range dari {Endy_pilih_dataset}: {Endy_jangkauan}")
                                    print()
                                    continue
                                elif Endy_penyebaran == "2":
                                    print("=" * 10, "Menu Quartil", "=" * 10)
                                    print("[1.] Quartil ke-1")
                                    print("[2.] Quartil ke-2")
                                    print("[3.] Quartil ke-3")
                                    print("[4.] All")
                                    print()
                                    
                                    Endy_pilih_quartil = input("Masukkan Pilihan Menu Quartil: ")
                                    if Endy_pilih_quartil == "1":
                                        Endy_Q1 = Endy_kalstat.Endy_kuartil(Endy_pilih_dataset, 1)
                                        print(f"Q1 dari {Endy_pilih_dataset}: {Endy_Q1}")
                                        print()
                                        continue
                                    elif Endy_pilih_quartil == "2":
                                        Endy_Q2 = Endy_kalstat.Endy_kuartil(Endy_pilih_dataset, 2)
                                        print(f"Q2 dari {Endy_pilih_dataset}: {Endy_Q2}")
                                        print()
                                        continue
                                    elif Endy_pilih_quartil == "3":
                                        Endy_Q3 = Endy_kalstat.Endy_kuartil(Endy_pilih_dataset, 3)
                                        print(f"Q3 dari {Endy_pilih_dataset}: {Endy_Q3}")
                                        print()
                                        continue
                                    elif Endy_pilih_quartil == "4":
                                        Endy_All = Endy_kalstat.Endy_kuartil_semua(Endy_pilih_dataset)
                                        print(f"Q1-3 dari {Endy_pilih_dataset}: {Endy_All}")
                                        print()
                                        continue
                                    else:
                                        print("Pilihan Menu tidak tersedia!!")
                                        print()
                                        continue
                                        
                                elif Endy_penyebaran == "3":
                                    Endy_interQ = Endy_kalstat.Endy_IQR(Endy_pilih_dataset)
                                    print(f"Inter Quartil dari {Endy_pilih_dataset}: {Endy_interQ}")
                                    print()
                                    continue
                                elif Endy_penyebaran == "4":
                                    Endy_var = Endy_kalstat.Endy_varian(Endy_pilih_dataset)
                                    print(f"Varians dari {Endy_pilih_dataset}: {Endy_var}")
                                    print()
                                    continue
                                elif Endy_penyebaran == "5":
                                    Endy_sd = Endy_kalstat.Endy_standar_deviasi(Endy_pilih_dataset)
                                    print(f"Standar Deviasi dari {Endy_pilih_dataset}: {Endy_sd}")
                                    print()
                                    continue
                                elif Endy_penyebaran == "6":
                                    Endy_penyebaran_all = {
                                        "Range": Endy_kalstat.Endy_range(Endy_pilih_dataset),
                                        "Quartil 1": Endy_kalstat.Endy_kuartil(Endy_pilih_dataset, 1),
                                        "Quartil 2": Endy_kalstat.Endy_kuartil(Endy_pilih_dataset, 2),
                                        "Quartil 3": Endy_kalstat.Endy_kuartil(Endy_pilih_dataset, 3),
                                        "IQR": Endy_kalstat.Endy_IQR(Endy_pilih_dataset),
                                        "Varian": Endy_kalstat.Endy_varian(Endy_pilih_dataset),
                                        "Standar Deviasi": Endy_kalstat.Endy_standar_deviasi(Endy_pilih_dataset)
                                    }

                                    for Endy_nama, Endy_isi in Endy_penyebaran_all.items():
                                        print(f"{Endy_nama}: {Endy_isi}")

                                    print()
                                    break
                                else:
                                    print("Pilihan Menu Tidak Tersedia!!")
                                    print()
                                    continue

        elif Endy_pilihan == 7:
            print("=" * 50)
            print("=" * 10, "Menampilkan Summary Statistics", "=" * 11)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Data base anda masih kosong!!")
                print()
                continue
            else:
                while True:
                    print("=" * 10, "Daftar Dataset", "=" * 10)
                    for Endy_nama in Endy_kalstat.Endy_data_base:
                        print(f"- {Endy_nama}")
    
                    print()
                    Endy_pilih_dataset = input("Masukkan nama dataset: ")
                    if Endy_pilih_dataset not in Endy_kalstat.Endy_data_base:
                        print("Dataset tidak ditemukan!!")
                        print()
                        continue
                    else:
                        if Endy_kalstat.Endy_panjang_data(Endy_pilih_dataset) == 0:
                            print("Isi Dataset Masih Kosong!")
                            print()
                            break
                        else:
                            print(f"Berikut merupakan summary statistics dari {Endy_pilih_dataset}:")
                            Endy_kalstat.Endy_summary(Endy_pilih_dataset)
                            print()
                            break
            
            
        elif Endy_pilihan == 8:
            print("=" * 50)
            print("=" * 11, "Menampilkan Korelasi Pearson", "=" * 11)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Database Anda Masih Kosong!")
                print()
                continue
            
            while True:
                print("=" * 10, "Tentukan Variabel X", "=" * 10)
                for Endy_namax in Endy_kalstat.Endy_data_base:
                    print(f"- {Endy_namax}")
                print()
                Endy_pilih_x = input("Masukkan Nama Variabel X Anda: ")
                if Endy_pilih_x not in Endy_kalstat.Endy_data_base:
                    print("Nama Dataset Tidak ditemukan!")
                    print()
                    continue

                print("=" * 10, "Tentukan Variabel Y", "=" * 10)
                for Endy_namay in Endy_kalstat.Endy_data_base:
                    print(f"- {Endy_namay}")
                print()
                Endy_pilih_y = input("Masukkan Nama Variabel Y Anda: ")
                if Endy_pilih_y not in Endy_kalstat.Endy_data_base:
                    print("Nama Dataset Tidak Ditemukan!")
                    print()
                    continue

                if Endy_pilih_x == Endy_pilih_y:
                    print("Kedua Dataset Tidak Boleh Sama!")
                    print()
                    continue

                Endy_panjang_x = Endy_kalstat.Endy_panjang_data(Endy_pilih_x)
                Endy_panjang_y = Endy_kalstat.Endy_panjang_data(Endy_pilih_y)
                if Endy_panjang_x == 0:
                    print("Isi Dataset X Tidak Boleh Kosong!")
                    print()
                    break
                elif Endy_panjang_y == 0:
                    print("Isi Dataset Y Tidak Boleh Kosong!")
                    print()
                    break
                elif Endy_panjang_x != Endy_panjang_y:
                    print("Panjang Isi Dataset X dan Y Harus Sama!")
                    print()
                    break
                else:
                    Endy_hasil_korelasi = Endy_kalstat.Endy_korelasi(Endy_pilih_x, Endy_pilih_y)
                    print(f"Hasil Hitung Korelasi Pearson: {Endy_hasil_korelasi}")
                    print()
                    break
    
        elif Endy_pilihan == 9:
            print("=" * 50)
            print("=" * 4, "Menampilkan Model Regresi Linear Sederhana", "=" * 4)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Database Anda Masih Kosong!")
                print()
                continue
            
            while True:
                print("=" * 10, "Pilih Variabel X", "=" * 10)
                for Endy_namax in Endy_kalstat.Endy_data_base:
                    print(f"- {Endy_namax}")
                print()
                Endy_pilih_x = input("Masukkan Nama Variabel X Anda: ")
                if Endy_pilih_x not in Endy_kalstat.Endy_data_base:
                    print("Nama Dataset Tidak Ditemukan!")
                    print()
                    continue

                print("=" * 10, "Pilih Variabel Y", "=" * 10)
                for Endy_namay in Endy_kalstat.Endy_data_base:
                    print(f"- {Endy_namay}")
                print()
                Endy_pilih_y = input("Masukkan Nama Variabel Y Anda: ")
                if Endy_pilih_y not in Endy_kalstat.Endy_data_base:
                    print("Nama Dataset Tidak Ditemukan: ")
                    print()
                    continue

                if Endy_pilih_x == Endy_pilih_y:
                    print("Kedua Dataset Tidak Boleh Sama!")
                    print()
                    continue

                Endy_panjang_x = Endy_kalstat.Endy_panjang_data(Endy_pilih_x)
                Endy_panjang_y = Endy_kalstat.Endy_panjang_data(Endy_pilih_y)

                if Endy_panjang_x == 0:
                    print("Isi Dataset X Tidak Boleh Kosong!")
                    print()
                    break
                elif Endy_panjang_y == 0:
                    print("Isi Dataset Y Tidak Boleh Kosong!")
                    print()
                    break
                elif Endy_panjang_x != Endy_panjang_y:
                    print("Panjang Isi Dataset X dan Y Harus Sama!")
                    print()
                    break
                else:
                    Endy_a, Endy_b = Endy_kalstat.Endy_model_regresi(Endy_pilih_x, Endy_pilih_y)
                    print(f"Model Regresi Yang Di Peroleh: {Endy_a} + {Endy_b}X")
                    break
    
        elif Endy_pilihan == 10:
            print("=" * 50)
            print("=" * 7, "Menampilkan Koefisien Determinasi", "=" * 8)
            print("=" * 50)
            print()

            if len(Endy_kalstat.Endy_data_base) == 0:
                print("Database Anda Masih Kosong!")
                print()
                continue

            while True:
                print("=" * 10, "Pilih Variabel X", "=" * 10)
                for Endy_namax in Endy_kalstat.Endy_data_base:
                    print(f"- {Endy_namax}")
                print()
                Endy_pilih_x = input("Masukkan Nama Variabel X Anda: ")
                if Endy_pilih_x not in Endy_kalstat.Endy_data_base:
                    print("Nama Dataset Tidak Ditemukan!")
                    print()
                    continue

                print("=" * 10, "Pilih Variabel Y", "=" * 10)
                for Endy_namay in Endy_kalstat.Endy_data_base:
                    print(f"- {Endy_namay}")
                print()
                Endy_pilih_y = input("Masukkan Nama Variabel Y Anda: ")
                if Endy_pilih_y not in Endy_kalstat.Endy_data_base:
                    print("Nama Dataset Tidak Ditemukan: ")
                    print()
                    continue

                if Endy_pilih_x == Endy_pilih_y:
                    print("Kedua Dataset Tidak Boleh Sama!")
                    print()
                    continue

                Endy_panjang_x = Endy_kalstat.Endy_panjang_data(Endy_pilih_x)
                Endy_panjang_y = Endy_kalstat.Endy_panjang_data(Endy_pilih_y)

                if Endy_panjang_x == 0:
                    print("Isi Dataset X Tidak Boleh Kosong!")
                    print()
                    break
                elif Endy_panjang_y == 0:
                    print("Isi Dataset Y Tidak Boleh Kosong!")
                    print()
                    break
                elif Endy_panjang_x != Endy_panjang_y:
                    print("Panjang Isi Dataset X dan Y Harus Sama!")
                    print()
                    break
                else:
                    Endy_hasil_determinasi = Endy_kalstat.Endy_determinasi(Endy_pilih_x, Endy_pilih_y)
                    print(f"Hasil Hitung Koefisien Determinasi: {Endy_hasil_determinasi}")
                    break
    
        elif Endy_pilihan == 11:
            print("=" * 15, "TERIMAKASIH", "=" * 15)
            break

        else:
            print("Pilihan Tidak Tersedia!")
            print()
            continue

    except ValueError:
        print("Pilihan yang dimasukkan harus berupa angka!!")
        print()
        continue
```
