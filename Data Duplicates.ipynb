# Import Packages

```python
import pandas as Endy_pd
import numpy as Endy_np
from fast_ml.feature_selection import get_duplicate_features
```

# Import Data

```python
Endy_data = Endy_pd.DataFrame({
    'Visitor Name': ['Ivan', 'Adhitya', 'Christ', 'Ivan', 'Daniel'],
    'Age': [25, 23, 18, 25, 25],
    'Event Name': ['Harmony Fest 2020', 'Soundscape Live 2021', 'Nova Concert 2022', 'Skyline Music Show 2023', 'Unity Sound Fest 2026'],
    'Ticket Sold Year': [2020, 2021, 2022, 2023, 2026],
    'Event Year': [2020, 2021, 2022, 2023, 2026],
    'Ticket Type': ['Gold', 'Silver', 'Bronze', 'Silver', 'Silver'],
    'Ticket Price': [1000000.0, 750000.0, 500000.0, 750000.0, 750000.0],
    'Reward Point': [400, 350, 800, 0, 250]
})

Endy_data['Age'] = Endy_data['Age'].astype('Int64')
Endy_data['Ticket Sold Year'] = Endy_data['Ticket Sold Year'].astype('Int64')
Endy_data['Event Year'] = Endy_data['Event Year'].astype('Int64')
Endy_data['Ticket Price'] = Endy_data['Ticket Price'].astype(float)
Endy_data['Reward Point'] = Endy_data['Reward Point'].astype('Int64')
Endy_data
```

```python
Endy_data2 = Endy_pd.DataFrame({
    'Visitor Name': ['Ivan', 'Adhitya', 'Christ', 'Ivan', 'Daniel'],
    'Age': [25, 23, 18, 25, 25],
    'Event Name': ['Harmony Fest 2020', 'Soundscape Live 2021', 'Nova Concert 2022', 'Harmony Fest 2020', 'Unity Sound Fest 2026'],
    'Ticket Sold Year': [2020, 2021, 2022, 2020, 2026],
    'Event Year': [2020, 2021, 2022, 2020, 2026],
    'Ticket Type': ['Gold', 'Silver', 'Bronze', 'Gold', 'Silver'],
    'Ticket Price': [1000000.0, 750000.0, 500000.0, 1000000.0, 750000.0],
    'Reward Point': [400, 350, 800, 400, 250]
})

Endy_data2['Age'] = Endy_data2['Age'].astype('Int64')
Endy_data2['Ticket Sold Year'] = Endy_data2['Ticket Sold Year'].astype('Int64')
Endy_data2['Event Year'] = Endy_data2['Event Year'].astype('Int64')
Endy_data2['Ticket Price'] = Endy_data2['Ticket Price'].astype(float)
Endy_data2['Reward Point'] = Endy_data2['Reward Point'].astype('Int64')
Endy_data2
```

# Olah Data

```python
#Menyisakan yang pertama
Endy_first = Endy_data.copy()
Endy_first.drop_duplicates(subset = 'Visitor Name', keep = 'first', inplace = True)
Endy_first
```

```python
#Menyisakan yang terakhir
Endy_last = Endy_data.copy()
Endy_last.drop_duplicates(subset = 'Visitor Name', keep = 'last', inplace = True)
Endy_last
```

```python
#Hapus semua yang duplikat
Endy_all = Endy_data.copy()
Endy_all.drop_duplicates(subset = 'Visitor Name', keep = False, inplace = True)
Endy_all
```

```python
#Hapus duplikasi yang barisnya sama semua
Endy_sama = Endy_data2.copy()
Endy_sama.drop_duplicates(inplace = True)
Endy_sama
```

```python
#Menangani Duplikasi Kolom
Endy_kolom = Endy_data2.copy()

kolom_duplikat = get_duplicate_features(Endy_kolom)
print('Duplikat kolom: \n')
print(kolom_duplikat)

list_duplikat = kolom_duplikat.loc[kolom_duplikat['Desc'] == 'Duplicate Values', 'feature2'].to_list()
print('\n List Kolom Duplikasi Values')
print(list_duplikat)

print('\n Banyaknya Kolom sebelum penghapusan kolom duplikasi values: ' + str(Endy_kolom.shape[1]))
Endy_kolom.drop(columns = list_duplikat, inplace = True)
print('Banyaknya kolom setelah penghapusan kolom duplikasi values: ' + str(Endy_kolom.shape[1]))
Endy_kolom
```

```python
#Menangani Duplikasi Kolom
Endy_index = Endy_data2.copy()

kolom_dupin = get_duplicate_features(Endy_index)
print('Duplikat kolom: \n')
print(kolom_dupin)

list_dupin = kolom_dupin.loc[kolom_dupin['Desc'] == 'Duplicate Index', 'feature2'].to_list()
print('\n List Kolom Duplikasi Index')
print(list_dupin)

print('\n Banyaknya Kolom sebelum penghapusan kolom duplikasi Index: ' + str(Endy_index.shape[1]))
Endy_index.drop(columns = list_dupin, inplace = True)
print('Banyaknya kolom setelah penghapusan kolom duplikasi Index: ' + str(Endy_index.shape[1]))
Endy_index
```
