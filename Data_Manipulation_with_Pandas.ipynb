```python
print("="*17, "Import Packages", "="*16)
import pandas as pd_endy ##Import package
```

```python
#Import Data CSV to Python
print("="*50)
print("="*12, "Import Data CSV to Python", "="*11)
print("="*50)

data_endy = pd_endy.read_csv("Data Manipulation with Pandas - data_karyawan.csv") ##Syntax to import CSV
print(data_endy)
```

```python
#To Preview Top and Bottom rows of data
print("="*50)
print("="*7, "To preview the top rows of the data", "="*6)
print("="*50)
print(data_endy.head()) ##Syntax to acsess

print()
print()

print("="*50)
print("="*5, "To preview the bottom rows of the data", "="*5)
print("="*50)
print(data_endy.tail()) ##Syntax to acsess
```

```python
#If we want to acsess some variable
print("="*50)
print("="*13, "To acsess some variable", "="*12)
print("="*50)
print(data_endy[['Umur', 'Gaji']]) ##Syntax to acsess
```

```python
#Add New Variable to Tabular
print("="*50)
print("="*11, "Add New Variable to Tabular", "="*10)
print("="*50)
data_endy['Gaji Bersih'] = data_endy['Gaji'] - data_endy['Pajak'] ##Syntax to add new variable
print(data_endy)
```

```python
#Change value on variable
print("="*50)
print("="*12, "Change Value on Variable", "="*12)
print("="*50)
data_endy["Status"] = data_endy["Status"].replace("single", "Belum Menikah")
data_endy["Status"] = data_endy["Status"].replace("married", "Sudah Menikah")
print(data_endy)
```

```python
#If we want to remove variable
print("="*50)
print("="*17, "Remove Variable", "="*16)
print("="*50)

data_endy = data_endy.drop("Pajak", axis = 1) #Syntax for remove variable 'Pajak'
print(data_endy)
```

```python
#Mean according to value of variable
print("="*50)
print("="*7, "Mean according to value of variable", "="*6)
print("="*50)

print(data_endy.groupby("Divisi")["Gaji"].mean())
```
