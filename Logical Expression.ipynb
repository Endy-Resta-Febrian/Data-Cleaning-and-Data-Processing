## Import Packages

```python
import pandas as Endy_pd
```

## Import Data

```python
Endy_data = Endy_pd.read_csv('Online Retail Data.csv', header = 0)
Endy_data
```

## Query Dengan Logical Expression

```python
#Mengakses data yang mengandung sebuah kata dan kondisi lainnya
Endy_data[(Endy_data['product_name'].str.contains('TEA')) & (Endy_data['quantity'] == 1)]
```

```python
#Mengakses data hanya mengandung suatu kata tertentu
Endy_data['product_name'].str.contains('RED')
```

```python
#Mengakses data yang mengandung sebuah kata menggunakan or
Endy_data[(Endy_data['product_name'].str.contains('TEA')) | (Endy_data['product_name'].str.contains('COFFE'))]
```

```python
#Mengakses data yang mengandung sebuah kata menggunakan or dan dalam kondisi tertentu
Endy_data[
    ((Endy_data['product_name'].str.contains('TEA')) | (Endy_data['product_name'].str.contains('COFFE')))
    & (Endy_data['price'] > 7)]
```

```python
#Akses beberapa kolom
Endy_data[['product_name', 'price']]
```
