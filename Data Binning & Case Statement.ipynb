# Import Packages

```python
import pandas as Endy_pd
import numpy as Endy_np
```

# Import Data CSV

```python
Endy_data = Endy_pd.read_csv('Online Retail Data.csv', header = 0)
Endy_data
```

```python
Endy_data = Endy_data[Endy_data['price'].between(0, 100, inclusive = 'right')].reset_index(drop = True)
```

# Binning Data

```python
#Membuat banyak bins tertentu dengan panjang yang sama
Endy_df1 = Endy_data.copy()
kategori = ['Murah', 'Standar', 'Mahal']
Endy_df1['Bin 1'] = Endy_pd.cut(Endy_df1['price'], 3, labels = kategori, include_lowest = True)
Endy_df1
```

```python
#Kita Tentukan bins nya
Endy_df2 = Endy_data.copy()
jumlah_bin = Endy_np.linspace(min(Endy_data['price']), max(Endy_data['price']), 4)
kategori = ['Murah', 'Standar', 'Mahal']
Endy_df2['Bin 2'] = Endy_pd.cut(Endy_df2['price'], jumlah_bin, labels = kategori, include_lowest = True)
Endy_df2
```

```python
#Menggunakan bins tapi kita tentukan jaraknya
Endy_df3 = Endy_data.copy()

jarak_interval = Endy_pd.interval_range(start = min(Endy_data['price']), freq = 10, end = max(Endy_data['price']))
Endy_df3['Bin 3'] = Endy_pd.cut(Endy_data['price'], bins = jarak_interval, include_lowest = True)
Endy_df3
```

```python
#Membuat bin tapi interval nya kita tentukan
Endy_df4 = Endy_data.copy()
batas_bin = [0, 30, 80, 100]
kategori = ['Murah', 'Standar', 'Mahal']
Endy_df4['Bin 4'] = Endy_pd.cut(Endy_data['price'], bins = batas_bin, labels = kategori, include_lowest = True)
Endy_df4
```

```python
#Menentukan banyak data dalam bin
Endy_df5 = Endy_data.copy()
kategori = ['Murah', 'Standar', 'Mahal']
Endy_df5['Bin 5'] = Endy_pd.qcut(Endy_data['price'], q = 3, labels = kategori)
Endy_df5
```
